# Employee Performance Analysis using Python

This beginner-friendly project analyzes employee information before building a Machine Learning model. The dataset contains working hours, training hours, experience, and performance score.

**Goal:** Understand which factors are related to employee performance using basic data analysis and visualization.

## 1. Import Libraries

We use Pandas for data handling, NumPy for numerical operations, Matplotlib and Seaborn for visualization.

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

sns.set_theme(style='whitegrid')
```

## 2. Load the Dataset

The dataset is stored in `employee_performance.csv`.

```python
df = pd.read_csv('employee_performance.csv')
df.head()
```

## 3. Understand the Data

```python
print('Dataset shape:', df.shape)
print('\nColumn information:')
df.info()

print('\nSummary statistics:')
df.describe()
```

## 4. Check Missing Values and Duplicates

Before analysis, we check whether any values are missing or duplicated.

```python
print('Missing values:')
print(df.isnull().sum())

print('\nDuplicate rows:', df.duplicated().sum())
```

## 5. Average Performance

The average performance score gives us a quick idea of the overall employee performance.

```python
average_performance = df['Performance_Score'].mean()
print(f'Average performance score: {average_performance:.2f}')
```

## 6. Working Hours vs Performance

This scatter plot helps us see whether employees who work more hours tend to have higher performance scores.

```python
plt.figure(figsize=(8, 5))
sns.scatterplot(data=df, x='Working_Hours', y='Performance_Score', s=70)
plt.title('Working Hours vs Performance Score')
plt.xlabel('Working Hours per Week')
plt.ylabel('Performance Score')
plt.show()
```

## 7. Training Hours vs Performance

Training may help employees improve their skills, so we check its relationship with performance.

```python
plt.figure(figsize=(8, 5))
sns.scatterplot(data=df, x='Training_Hours', y='Performance_Score', s=70)
plt.title('Training Hours vs Performance Score')
plt.xlabel('Training Hours')
plt.ylabel('Performance Score')
plt.show()
```

## 8. Experience vs Performance

We also check whether more years of experience are associated with better performance.

```python
plt.figure(figsize=(8, 5))
sns.scatterplot(data=df, x='Experience_Years', y='Performance_Score', s=70)
plt.title('Experience vs Performance Score')
plt.xlabel('Experience (Years)')
plt.ylabel('Performance Score')
plt.show()
```

## 9. Correlation Analysis

Correlation shows how strongly two numerical variables move together. Values closer to +1 indicate a strong positive relationship, while values closer to -1 indicate a strong negative relationship.

```python
correlation = df.drop(columns=['Employee_ID']).corr()
correlation
```

```python
plt.figure(figsize=(8, 6))
sns.heatmap(correlation, annot=True, cmap='coolwarm', fmt='.2f')
plt.title('Correlation Matrix')
plt.show()
```

## 10. Main Findings

- Working hours, training hours, and experience all show positive relationships with performance in this sample.
- Training hours are useful to examine because training can improve employee skills.
- Experience also appears to be related to performance.
- The analysis is exploratory and shows relationships, not proof of cause and effect.
- The cleaned numerical features can be used as inputs for a future Machine Learning regression model.

## 11. Conclusion

The analysis gives a simple picture of the factors associated with employee performance. In this dataset, employees with more working hours, training hours, and experience generally have higher performance scores. These results can be used as a starting point for building a Machine Learning model that predicts employee performance.
